```python
def get_all_h5ad_files(directory):
    h5ad_files = []
    for root, _, files in os.walk(directory):
        for file in files:
            if file.endswith(".h5ad"):
                h5ad_files.append(os.path.join(root, file))
    return h5ad_files
```

```python
scH5ADfiles = get_all_h5ad_files(DATAPATH)
```
